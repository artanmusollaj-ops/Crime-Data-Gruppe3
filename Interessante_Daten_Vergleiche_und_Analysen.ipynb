# 1. Kriminalität in Städten, Vororten und ländlichen Gebieten

Das passt besonders gut zu deiner Frage nach kleineren US-Städten.

Mögliche Hypothesen:

Großstädte haben höhere Raub- und Taschendiebstahlraten.
Vororte haben relativ mehr Kfz-Diebstahl und Wohnungseinbruch.
Ländliche Gebiete haben andere Deliktprofile, aber wegen kleiner Fallzahlen stärkere Schwankungen.
Universitätsstädte wie Chapel Hill oder Bloomington weisen saisonale Effekte zum Semesterbeginn auf.

Verwende nicht die absoluten Fallzahlen, sondern beispielsweise:

$$ \text{Kriminalitätsrate}_{it} = \frac{\text{Fälle}_{it}}{\text{Bevölkerung}_{it}} \times 100.000 $$

Für Einbruch können Haushalte und für Kfz-Diebstahl registrierte Fahrzeuge sinnvollere Nenner sein als Einwohner.

Als Modell eignet sich eine Negative-Binomial-Regression:

$$ \log(E[\text{Fälle}_{it}]) = \beta X_{it}+\log(\text{Bevölkerung}_{it}) $$

Der logarithmierte Bevölkerungswert wird dabei als Offset verwendet.

Bildungsstand
Urbanisierungsgrad
Wohnkosten
Tourismusübernachtungen
Pendleraufkommen
Polizeidichte
Anzahl von Bars, Clubs oder ÖPNV-Stationen

Interessante Fragestellungen wären:

Hängt Eigentumskriminalität stärker mit Arbeitslosigkeit oder Bevölkerungsdichte zusammen?
Bleibt der Zusammenhang bestehen, wenn man Urbanisierung und Altersstruktur kontrolliert?
Erklärt Tourismus erhöhte Taschendiebstahlraten besser als die Wohnbevölkerung?
Haben einkommensstarke Vororte weniger Gewaltkriminalität, aber nicht unbedingt weniger Kfz-Diebstahl?

Für den Einstieg eignen sich Spearman-Korrelationen besser als Pearson, weil Kriminalitätsraten häufig schief verteilt sind und Ausreißer enthalten.

Danach solltest du Partial Correlations oder Regressionen verwenden:

import statsmodels.formula.api as smf

model = smf.ols(
    "crime_rate ~ unemployment + income + population_density + young_adult_share",
    data=panel
).fit(cov_type="HC3")

print(model.summary())

Noch besser ist ein Panelmodell mit festen Effekten für Region und Jahr. Damit vergleichst du eine Region stärker mit ihrer eigenen Entwicklung als mit völlig unterschiedlichen Regionen.

# 4. Verschiebung der Kriminalitätsstruktur

Ein besonders gutes Thema wäre:

Geht klassische Eigentumskriminalität zurück, während Betrug und Cybercrime zunehmen?

Dafür berechnest du pro Region und Jahr den Anteil jeder Deliktgruppe an allen erfassten Delikten:

$$ p_{kit}=\frac{\text{Fälle der Kategorie }k}{\text{alle Fälle}} $$

Zusätzlich kannst du die Diversität des Deliktprofils über die Shannon-Entropie messen:

$$ H=-\sum_k p_k\log(p_k) $$

Eine hohe Entropie bedeutet, dass die Kriminalität auf viele Deliktarten verteilt ist; eine niedrige Entropie zeigt eine starke Konzentration auf wenige Kategorien.

Anschließend kannst du Städte nach ihrem Deliktprofil clustern, etwa in:

Gewalt-dominierte Gebiete
Eigentumsdelikt-dominierte Gebiete
Kfz-Delikt-Cluster
Betrugs-/Cybercrime-Cluster
Gebiete mit insgesamt geringer Belastung
# 5. Räumliche Konzentration und Spillover

Nicht nur „Wo ist die Rate hoch?“, sondern:

Sind hohe Raten räumlich geclustert?
Haben Nachbarregionen ähnliche Entwicklungen?
Verlagert sich Kriminalität nach einer Intervention in angrenzende Gebiete?
Konzentrieren sich 50 % der Fälle auf 5 % der Gebiete?

Geeignete Verfahren:

Global Moran’s I
Local Moran’s I/LISA
Getis-Ord Gi*
räumlich verzögerte Variablen
räumliche Panelmodelle
Lorenzkurve und Gini-Koeffizient der räumlichen Konzentration

Ein normaler Machine-Learning-Split ist hier gefährlich: Benachbarte Gebiete in Training und Test erzeugen Spatial Leakage. Besser ist eine räumliche Cross-Validation, bei der ganze Regionen gemeinsam zurückgehalten werden.

# 7. Aufklärungsquote und „Clearance Gap“

Falls Fälle und aufgeklärte Fälle vorhanden sind:

$$ \text{Aufklärungsquote} = \frac{\text{aufgeklärte Fälle}}{\text{registrierte Fälle}} $$

Interessant sind:

Unterschiede nach Deliktart
Entwicklung über die Zeit
Verhältnis zwischen Fallbelastung und Aufklärungsquote
Unterschiede zwischen vergleichbaren Städten
mögliche Sättigungseffekte bei hoher Fallbelastung

Kleine Regionen sollten in Funnel Plots dargestellt werden. Eine Quote von 100 % bei zwei Fällen ist nicht direkt mit 70 % bei 20.000 Fällen vergleichbar.

# 9. Predictive Use Cases

Sinnvoll ist eine Prognose aggregierter Fallzahlen, etwa für Personal- oder Ressourcenplanung:

Fälle pro Region und Monat
Calls for Service pro Wochentag und Schicht
Wahrscheinlichkeit ungewöhnlich hoher Fallbelastung
erwartete Zahl bestimmter Deliktgruppen

Weniger empfehlenswert ist die Vorhersage, welche konkrete Person wahrscheinlich Täter wird. Das ist ethisch, rechtlich und statistisch hochproblematisch. Ein gutes Projekt bleibt auf aggregierter Raum-Zeit-Ebene.

Meine drei stärksten Projektvorschläge
1. Internationales Stadt–Land-Projekt

„Crime profiles in urban, suburban and rural areas: Germany, Italy, France and the USA“

Enthält räumliche Visualisierung, Clustering, sozioökonomische Variablen und Panelregression.

2. Veränderung der Deliktstruktur

„From burglary to online fraud: structural changes in recorded crime“

Enthält Zeitreihen, Change Points, COVID-Effekt und Deliktanteile.

3. US-Vororte als hochauflösendes Fallbeispiel

„When and where does reported crime occur in American suburbs?“

Montgomery County, Cary, Chapel Hill und Bloomington mit Tageszeit, Saison, Raum, Wetter und Landnutzung.

Wichtigste methodische Fallstricke
Polizeilich registrierte Kriminalität ist nicht die tatsächliche Gesamtkriminalität.
Mehr Anzeigen können wie mehr Kriminalität aussehen.
Mehr Polizeikontrollen erzeugen bei Kontrolldelikten mehr registrierte Fälle.
Deliktdefinitionen unterscheiden sich zwischen Ländern.
Korrelation bedeutet keine Kausalität.
Arbeitslosigkeit, Einkommen und Urbanisierung hängen untereinander zusammen.
Kleine Gemeinden benötigen Konfidenzintervalle oder mehrjährige Mittelwerte.
Rohzahlen dürfen nicht ohne passende Bevölkerungs- oder Risikobasis verglichen werden.
US-Behörden melden nicht in allen Jahren vollständig an NIBRS; die Abdeckung muss als Qualitätsvariable berücksichtigt werden. NIBRS deckte im Mai 2024 etwa 82 % der US-Bevölkerung ab. Bureau of Justice Statistics

Am stärksten wäre wahrscheinlich die Kombination aus internationalem Panelvergleich auf Kreis-/Provinzebene und einem detaillierten US-Vorort-Fallbeispiel auf Incident-Ebene. Dadurch demonstrierst du ETL, Datenmodellierung, Geodaten, Statistik, Zeitreihen und Machine Learning in einem zusammenhängenden Projekt.
