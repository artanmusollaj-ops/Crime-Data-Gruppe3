```python
import io
import requests
import pandas as pd

url = (
    "https://data.stadt-zuerich.ch/dataset/"
    "ktzh_pks_straftaten_tatbestandgruppe_gemeinden_stadtkreise/"
    "download/KTZH_00001202_00003600.csv"
)

headers = {
    "User-Agent": "CrimeDataResearch/1.0 (wissenschaftliches Projekt)"
}

response = requests.get(url, headers=headers, timeout=60)
response.raise_for_status()

df = pd.read_csv(
    io.BytesIO(response.content),
    sep=None,
    engine="python"
)

df.head()
```
